# RF downconversion demo
減衰RF信号をreferenceでミキシングしてI/Qから位相差を推定する例です。

```python
import numpy as np
import matplotlib.pyplot as plt

from signal_eval import SimulationConfig, run_simulation
```

```python
cfg = SimulationConfig(
    duration_s=10e-6,
    sample_rate_hz=200e6,
    rf_freq_hz=20e6,
    ref_freq_hz=20e6,
    damping_tau_s=2e-6,
    rf_phase_rad=np.deg2rad(35.0),
)
res = run_simulation(cfg)
actual = np.arctan2(np.sin(cfg.rf_phase_rad - cfg.ref_phase_rad), np.cos(cfg.rf_phase_rad - cfg.ref_phase_rad))
print(f'actual phase difference [rad]:    {actual:.6f}')
print(f'estimated phase difference [rad]: {res["estimated_phase_rad"]:.6f}')
```

```python
t_us = res['t'] * 1e6

fig, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True)
axes[0].plot(t_us, res['rf_signal'])
axes[0].set_ylabel('RF amplitude')
axes[0].set_title('Damped RF signal')

axes[1].plot(t_us, res['i_baseband'], label='I')
axes[1].plot(t_us, res['q_baseband'], label='Q')
axes[1].set_ylabel('Baseband')
axes[1].legend()

axes[2].plot(res['i_baseband'], res['q_baseband'])
axes[2].set_xlabel('I')
axes[2].set_ylabel('Q')
axes[2].set_title('IQ trajectory')

plt.tight_layout()
plt.show()
```
